#### Видно, что самым проблемным компрессором является RLE, что логично: данный компрессор не эффективен для данных, где встречается малое количество повторяющихся последовательностей байтов, и следовательно enwik7, ru_text, exe_file не сжимаются. Объяснить неэффективность других компрессоров можно также через неэффективную структуру алгоритма для определенных типов файлов, а также из за размера метаданных

### Лучшие компрессоры для каждого файла

```python
files = df['Data_File'].unique()

for file in files:
    file_df = df[df['Data_File'] == file]
    best = file_df[file_df['Compress_factor'] == file_df['Compress_factor'].max()]
    print(f'Лучший компрессор для {file} - {best['Compressor'].values}, compress_factor = {best['Compress_factor'].values}') 
    
    
    
```

### Худшие компрессоры для каждого файла

```python
files = df['Data_File'].unique()

for file in files:
    file_df = df[df['Data_File'] == file]
    best = file_df[file_df['Compress_factor'] == file_df['Compress_factor'].min()]
    print(f'Худший компрессор для {file} - {best['Compressor'].values}, compress_factor = {best['Compress_factor'].values}') 
```

### Визуализация на диаграммах

```python
path = './visualization/'
```

```python
plt.rcParams['font.size'] = 9
```

```python
# enwik
fig, ax = plt.subplots(figsize = (20, 20))
i = 0
compressors = df['Compressor'].unique()
good_values = df[(df["Data_File"] == 'enwik7') & (df['Compress_factor'] > 1)]['Compress_factor'].values
good_compressors = df[(df["Data_File"] == 'enwik7') & (df['Compress_factor'] > 1)]['Compressor'].values
bad_values = df[(df["Data_File"] == 'enwik7') & (df['Compress_factor'] <= 1)]['Compress_factor'].values
bad_compressors = df[(df["Data_File"] == 'enwik7') & (df['Compress_factor'] <= 1)]['Compressor'].values
ax.set_title('enwik7')
ax.set_ylabel('Compress Factor')
ax.set_xlabel('Compressors')
ax.bar(good_compressors, good_values, label = 'Compress factor > 1')
ax.bar(bad_compressors, bad_values, color = 'r', label = 'Compress factor <= 1')
ax.axhline(y=1, color='r', linestyle=':', linewidth=2)
ax.legend()
fig.savefig(path + 'enwik7.png')
```
